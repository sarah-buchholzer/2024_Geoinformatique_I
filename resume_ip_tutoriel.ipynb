```python
#vérifier le type d'un objet
a = 10
b = 'Hello'
print(type(a), type(b))
print(type(a) is int)
print(type(a) is str)
```

String (chaîne de caractères)





```python
# Concaténation de chaînes de caractères
text1,text2,text3,text4 = 'Introduction','to','Python','course'
print(text1+text2+text3+text4)
```

```python
#ajouter des espaces entre les mots
print(text1+' '+text2+' '+text3+' '+text4)
```

```python
#positions
print(text1[0],text2[1],text3[2],text4[3])
```

```python
# Tranchage de chaînes
print(text1[:5],text1[-5:],text1[:5]+text3[0:2])
#[:5]: 5 premiers, [5:]: 5 derniers
```

```python
# Formatage des chaînes de caractères
#f strings permettent de formater facilement les données
print(f'La variable a a pour type {type(a)} et pour valeur {a}')
print(f'La variable b est de type {type(b)} et de valeur {b}')
```

## Tutoriel 2: Structure de contrôle
Les instructions conditionnelles

```python
x = 100
if x > 0:
    print('Nombre positif')
elif x < 0:
    print('Nombre négatif')
else:
    print ('Zéro!')
```

Instructions de boucle


```python
# [range(5)==[0,1,2,3,4]]
for i in range(5):
    print(i)
```

```python
# itérer sur une liste que nous créons, et accéder à la fois aux indices et aux éléments avec enumerate()
for index,pet in enumerate(['dog', 'cat', 'fish']):
    print(index, pet, len(pet))
```

```python
# faire une boucle
count = 0
while count < 10:
    count += 1
print(count)
```

Instructions de contrôle de boucle

```python
for i in range(1, 10):
    if i == 5:
        print('Condition satisfaite')
        break
    print(i)  # break arrete tout et sort de la fonction
```

```python
for i in range(1, 10):
    if i == 5:
        print('Condition satisfaite')
        continue
        print("Quoi qu'il en soit. De toute façon, je ne serai pas imprimé.")
    print(i) #continue sort de la boucle
```

```python
for i in range(1, 10):
    if i == 5:
        print('Condition satisfaite')
        pass
    print(i) #pass ne fait rien
```
